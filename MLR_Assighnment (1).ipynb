Handle the outlier

```python
df1=df.copy()
for i in df1['Price']:
    q1 = np.quantile(df1.Price,0.25)
    q3 = np.quantile(df1.Price,0.75)
    med = np.median(df1.Price)
    iqr = q3 - q1
    upper_bound = q3+(1.5*iqr)
    lower_bound = q1-(1.5*iqr)
    if i > upper_bound or i < lower_bound:
        df1['Price'] = df1['Price'].replace(i, np.median(df1['Price']))
sns.boxplot(df1['Price'])
plt.title('Price after median imputation')
plt.show()
```

```python
for i in df1['Age_08_04']:
    q1 = np.quantile(df1.Age_08_04,0.25)
    q3 = np.quantile(df1.Age_08_04,0.75)
    med = np.median(df1.Age_08_04)
    iqr = q3 - q1
    upper_bound = q3+(1.5*iqr)
    lower_bound = q1-(1.5*iqr)
    if i > upper_bound or i < lower_bound:
        df1['Age_08_04'] = df1['Age_08_04'].replace(i, np.median(df1['Age_08_04']))
sns.boxplot(df1['Age_08_04'])
plt.title('Age after median imputation')
```

```python
for i in df1['KM']:
    q1 = np.quantile(df1.KM,0.25)
    q3 = np.quantile(df1.KM,0.75)
    med = np.median(df1.KM)
    iqr = q3 - q1
    upper_bound = q3+(1.5*iqr)
    lower_bound = q1-(1.5*iqr)
    if i > upper_bound or i < lower_bound:
        df1['KM'] = df1['KM'].replace(i, np.median(df1['KM']))
sns.boxplot(df1['KM'])
plt.title('KM after median imputation')
plt.show()
```

```python
for i in df1['Weight']:
    q1 = np.quantile(df1.Weight,0.25)
    q3 = np.quantile(df1.Weight,0.75)
    med = np.median(df1.Weight)
    iqr = q3 - q1
    upper_bound = q3+(1.5*iqr)
    lower_bound = q1-(1.5*iqr)
    if i > upper_bound or i < lower_bound:
        df1['Weight'] = df1['Weight'].replace(i, np.median(df1['Weight']))
sns.boxplot(df1['Weight'])
plt.title('Weight after median imputation')
plt.show()
```

```python
df2=df.copy()
def drop_outliers(column):
    q1 = column.quantile(0.25)
    q3 = column.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    return column[(column > lower_bound) & (column < upper_bound)]

# Remove outliers from each column
for col in df.columns:
    df2[col] = drop_outliers(df2[col])
```

1.What is Normalization & Standardization and how is it helpful?



Normalization and standardization are two techniques used to transform data in a more consistent and uniform format. Let’s explore each of them:


Normalization:

Objective: Normalize data so that each value falls between 0 and 1.
Formula:

For a value xi in the dataset:xnew​=xmax​−xmin​xi​−xmin​​

Where:

xi: The original value.
xmin: The minimum value in the dataset.
xmax: The maximum value in the dataset.
Standardization:

Objective: Transform data to have a mean of 0 and a standard deviation of 1.
Formula:

For a value xi in the dataset:xnew​=sxi​−xˉ​

Where:

xi: The original value.
x: The sample mean.
s: The sample standard deviation.

2.What techniques can be used to address multicollinearity in multiple linear regression?

Certainly! Multicollinearity occurs when independent variables in a regression model are highly correlated, which can cause problems during model fitting and interpretation. Here are some techniques to address multicollinearity in multiple linear regression:

Feature Selection:
Remove one or more of the highly correlated variables.
Prioritize variables based on their relevance to the research question or domain knowledge.
Feature Engineering:
Create new features by linearly combining existing predictor variables (e.g., adding or subtracting them).
Transform variables (e.g., taking logarithms) to reduce correlation.
Regularization Techniques:
Ridge Regression (L2 regularization):
Adds a penalty term to the regression coefficients.
Shrinks coefficients toward zero, reducing their sensitivity to multicollinearity.
Lasso Regression (L1 regularization):
Encourages sparsity by setting some coefficients exactly to zero.
Automatically selects relevant features and reduces multicollinearity.
